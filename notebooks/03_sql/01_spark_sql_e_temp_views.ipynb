## Task 1: deixar os dados do projeto consultáveis em SQL

```python
import sys
sys.path.insert(0, "../../src")

from pyspark.sql import SparkSession, functions as F

spark = (
    SparkSession.builder
    .appName("arena-sql")
    .master("local[*]")
    .config("spark.sql.warehouse.dir", "../../data/warehouse")
    .getOrCreate()
)

RAW = "../../data/raw"

spark
```

## As tabelas do projeto

As funções de `src/wc_pipeline/transformacoes.py`, do fim da aula 2:

```python
from wc_pipeline.schemas import MATCHES
from wc_pipeline.transformacoes import limpar_partidas, partidas_por_selecao, resumo_por_selecao

partidas = limpar_partidas(
    spark.read.option("header", "true").schema(MATCHES).csv(f"{RAW}/WorldCupMatches.csv")
)
por_selecao = partidas_por_selecao(partidas)
resumo = resumo_por_selecao(por_selecao)

print(partidas.count())
print(por_selecao.count())
print(resumo.count())
```

A célula abaixo já vem pronta: as escalações e as Copas, com o mesmo código do vídeo 2.4.

```python
# vídeo 2.4: escalações e Copas
from wc_pipeline.schemas import PLAYERS, WORLD_CUPS

escalacoes = (
    spark.read.option("header", "true").schema(PLAYERS).csv(f"{RAW}/WorldCupPlayers.csv")
    .select(
        F.col("MatchID").alias("match_id"),
        F.col("Team Initials").alias("sigla"),
        F.col("Player Name").alias("jogador"),
        F.col("Position").alias("posicao"),
    )
)

copas = (
    spark.read.option("header", "true").schema(WORLD_CUPS).csv(f"{RAW}/WorldCups.csv")
    .select(
        F.col("Year").alias("ano"),
        F.col("Country").alias("sede"),
        F.col("Winner").alias("campeao"),
        F.col("MatchesPlayed").alias("partidas_oficiais"),
        F.col("GoalsScored").alias("gols_oficiais"),
    )
)
```

## Views: dando um nome ao DataFrame

`createOrReplaceTempView` registra um nome para usar no `FROM`. A view não copia nem grava dado: é um nome para o
plano, e some quando a sessão termina.

```python
partidas.createOrReplaceTempView("partidas")
por_selecao.createOrReplaceTempView("por_selecao")
resumo.createOrReplaceTempView("resumo")
escalacoes.createOrReplaceTempView("escalacoes")
copas.createOrReplaceTempView("copas")

spark.sql("SHOW TABLES").show()
```

## Consultando com `spark.sql`

`spark.sql("...")` devolve **um DataFrame**.

```python
spark.sql("""
    SELECT ano, mandante, gols_mandante, gols_visitante, visitante
    FROM partidas
    WHERE gols_mandante >= 5
    ORDER BY ano
    LIMIT 5
""").show()
```

## A mesma pergunta, duas formas

```python
via_api = (
    partidas.groupBy("ano")
    .agg(F.sum("total_gols").alias("gols"))
    .orderBy("ano")
)

via_sql = spark.sql("""
    SELECT ano, SUM(total_gols) AS gols
    FROM partidas
    GROUP BY ano
    ORDER BY ano
""")

via_api.show(3)
via_sql.show(3)
```

## Misturando as duas formas

`spark.sql()` devolve um DataFrame: dá para começar em SQL e continuar em Python.

```python
(
    spark.sql("SELECT ano, selecao, gols_pro FROM resumo WHERE gols_pro >= 15")
    .groupBy("selecao")
    .agg(F.count("*").alias("copas_com_15_gols_ou_mais"))
    .orderBy(F.desc("copas_com_15_gols_ou_mais"))
    .show(5)
)
```

As cinco tabelas do projeto estão consultáveis em SQL.

## Para consultar depois

- **`createOrReplaceTempView`** substitui a view se o nome já existir. Existe também `createTempView`, que dá erro se o
  nome já estiver em uso.
- **Nomes com espaço em SQL:** use crases, por exemplo `` SELECT `Home Team Name` FROM tabela ``.
- **`spark.table("partidas")`** devolve a view como DataFrame, o caminho inverso do `createOrReplaceTempView`.
